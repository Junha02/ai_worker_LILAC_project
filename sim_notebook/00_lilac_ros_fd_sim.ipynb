## Runtime

- Vader5 left stick publishes `/vader5/latent_z`.
- Utterance text goes through `/lilac/apply_utterance`.
- The learned LILAC decoder publishes `/lilac/ee_delta_6d`.
- IK publishes `/sh5/qpos_cmd` and MuJoCo applies forward dynamics.
- MuJoCo publishes `/sh5/qpos_state` as robot/sim feedback.

````python
USE_VADER5 = True
INITIAL_UTTERANCE = "Pick up the cup and pour water into the bowl."

runtime = LILACROSSimRuntime(
    use_vader5        = USE_VADER5,
    initial_utterance = INITIAL_UTTERANCE,
    inference_hz      = 30,
    render_hz         = 30,
)
runtime.init_viewer()
runtime.print_graph()
````

````python
runtime.run()
````
